# 🪱 NeurWorm: C. elegans Connectome Exploration & Dynamics

This notebook explores the 302-neuron *Caenorhabditis elegans* connectome and simulates signal propagation using Leaky Integrate-and-Fire (LIF) dynamics.

```python
import numpy as np
import matplotlib.pyplot as plt
from connectome.loader import load_connectome, build_network
from connectome.neuron_groups import GROUP_LABELS

syn, gap, names = load_connectome()
print(f"Loaded {len(names)} neurons.")
print(f"Chemical synapses non-zero: {np.count_nonzero(syn)}")
print(f"Electrical gap junctions non-zero: {np.count_nonzero(gap)}")
```

## 1. Adjacency Matrix Visualization
We visualize the chemical synapse matrix and electrical gap junctions.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
im0 = axes[0].imshow(syn, cmap="viridis", aspect="auto")
axes[0].set_title("Chemical Synapses (Signed)")
plt.colorbar(im0, ax=axes[0])

im1 = axes[1].imshow(gap, cmap="plasma", aspect="auto")
axes[1].set_title("Electrical Gap Junctions (Symmetric)")
plt.colorbar(im1, ax=axes[1])
plt.tight_layout()
plt.show()
```

## 2. Sensory Stimulation Test
Inject current into chemosensory neuron `ASEL` (attractive food stimulus) and nociceptor `ASHL` (aversive stimulus), tracking membrane voltage propagation.

```python
net = build_network()
voltages = []
spikes_record = []

# 500 ms simulation (1000 steps of 0.5 ms)
for t in range(1000):
    if 100 <= t <= 400:
        net.inject("ASEL", 2.5)  # 2.5 nA food stimulus
    else:
        net.clear_external()
    spk = net.step()
    voltages.append(net.get_voltage_array())
    spikes_record.append(spk)

voltages = np.array(voltages)
print(f"Simulation complete. Voltage record shape: {voltages.shape}")
```
